```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv('Lectures/MLT-03 -  Supervised Learning/bioresponse.csv')
```

```python
data
```

```python
x = data.drop('Activity', axis=1)
y = data['Activity']

x.columns = x.columns.astype(str)

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=88)
```

```python
# small decision tree
from sklearn import metrics, tree

small_tree_classifier = tree.DecisionTreeClassifier(random_state=1, max_depth=1)
small_tree_classifier.fit(x_train, y_train)

y_small_tree_pred = small_tree_classifier.predict(x_test)
small_tree_accuracy = metrics.accuracy_score(y_test, y_small_tree_pred)
print(small_tree_accuracy)
```

```python
# deep decision tree
deep_tree_classifier = tree.DecisionTreeClassifier(random_state=1, max_depth=6)
deep_tree_classifier.fit(x_train, y_train)

y_deep_tree_pred = deep_tree_classifier.predict(x_test)
deep_tree_accuracy = metrics.accuracy_score(y_test, y_deep_tree_pred)
print(deep_tree_accuracy)
```

```python
# random forest on small trees
from sklearn.ensemble import RandomForestClassifier

small_rf_classifier = RandomForestClassifier(n_estimators = 100, max_depth = 5, random_state = 1)

small_rf_classifier.fit(x_train, y_train)

y_small_rf_pred = small_rf_classifier.predict(x_test)
small_rf_accuracy = metrics.accuracy_score(y_test, y_small_rf_pred)
print(small_rf_accuracy)
```

```python
# random forest on deep trees

deep_rf_classifier = RandomForestClassifier(n_estimators = 150, max_depth = 19, random_state = 1)

deep_rf_classifier.fit(x_train, y_train)

y_deep_rf_pred = deep_rf_classifier.predict(x_test)
deep_rf_accuracy = metrics.accuracy_score(y_test, y_deep_rf_pred)
print(deep_rf_accuracy)
```

2. Calculate the following metrics to check the quality of your models:
- precision;
- recall;

- F1-score;

```python
print("--- Precision ---")
print(f"Small Tree: {metrics.precision_score(y_test, y_small_tree_pred)}")
print(f"Deep Tree: {metrics.precision_score(y_test, y_deep_tree_pred)}")
print(f"Small RF: {metrics.precision_score(y_test, y_small_rf_pred)}")
print(f"Deep RF: {metrics.precision_score(y_test, y_deep_rf_pred)}")

print("\n--- Recall ---")
print(f"Small Tree: {metrics.recall_score(y_test, y_small_tree_pred)}")
print(f"Deep Tree: {metrics.recall_score(y_test, y_deep_tree_pred)}")
print(f"Small RF: {metrics.recall_score(y_test, y_small_rf_pred)}")
print(f"Deep RF: {metrics.recall_score(y_test, y_deep_rf_pred)}")

print("\n--- Accuracy ---")
print(f"Small Tree: {small_tree_accuracy}")
print(f"Deep Tree: {deep_tree_accuracy}")
print(f"Small RF: {small_rf_accuracy}")
print(f"Deep RF: {deep_rf_accuracy}")

print("\n--- F1 Score ---")
print(f"Small Tree: {metrics.f1_score(y_test, y_small_tree_pred)}")
print(f"Deep Tree: {metrics.f1_score(y_test, y_deep_tree_pred)}")
print(f"Small RF: {metrics.f1_score(y_test, y_small_rf_pred)}")
print(f"Deep RF: {metrics.f1_score(y_test, y_deep_rf_pred)}")

print("\n--- Log-loss ---")
print(f"Small Tree: {metrics.log_loss(y_test, small_tree_classifier.predict_proba(x_test))}")
print(f"Deep Tree: {metrics.log_loss(y_test, deep_tree_classifier.predict_proba(x_test))}")
print(f"Small RF: {metrics.log_loss(y_test, small_rf_classifier.predict_proba(x_test))}")
print(f"Deep RF: {metrics.log_loss(y_test, deep_rf_classifier.predict_proba(x_test))}")
```
